```python
from pathlib import Path
import json, sys, subprocess
from IPython.display import display, Image, Markdown
candidates = [Path.cwd(), Path.cwd()/'ML5_Deadline_5h', Path.cwd().parent/'ML5_Deadline_5h', Path.cwd()/'outputs'/'ML5_Deadline_5h', Path(r'C:\Users\84969\Documents\Codex\2026-10-12\referenced-chatgpt-conversation-this-is-an\outputs\ML5_Deadline_5h')]
BASE = next((p for p in candidates if (p/'run_deadline_ml5.py').exists()), None)
if BASE is None:
    raise FileNotFoundError('Mở thư mục ML5_Deadline_5h bằng File > Open Folder rồi chọn lại kernel.')
RESULTS = BASE/'results'
print('Thư mục kết quả:', RESULTS)
```

## Theo dõi tiến độ
Chạy lại ô dưới để đọc tiến độ mới nhất. Kết quả tổng hợp và biểu đồ xuất sau mỗi ma trận 18 lượt hoàn chỉnh. Không kết luận từ các dòng chưa hoàn tất.

```python
progress = RESULTS/'progress.json'
if progress.exists():
    display(json.loads(progress.read_text(encoding='utf-8')))
else:
    print('Chưa bắt đầu huấn luyện hoặc chưa có tiến độ được lưu.')
for name in ['stdout.log','stderr.log']:
    p = BASE/name
    if p.exists():
        print('\n'+name+'\n'+'\n'.join(p.read_text(encoding='utf-8',errors='replace').splitlines()[-15:]))
```

## Chỉ dùng khi cần tự khởi chạy

```python
RUN_TRAINING = False
CONFIRM_FULL_STOPPED = False
if RUN_TRAINING:
    if not CONFIRM_FULL_STOPPED:
        raise RuntimeError('Dừng lượt full trước rồi mới xác nhận CONFIRM_FULL_STOPPED=True.')
    subprocess.run([sys.executable,'-u',str(BASE/'run_deadline_ml5.py'),'--confirm-full-stopped','--hours','4'],cwd=BASE,check=True)
else:
    print('Chế độ chỉ đọc kết quả; không khởi chạy thêm.')
```
